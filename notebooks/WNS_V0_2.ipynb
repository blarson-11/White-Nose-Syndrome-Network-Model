# County-Level Stochastic SI Modeling of White-Nose Syndrome

## Project Overview

This project explores potential disease spread through a network of U.S. counties using stochastic modeling and Monte Carlo simulation. 

Each county is represented as a node in an undirected graph, with edges representing the geographic connection between neighboring counties. 

A stochastic SI (Susceptible-Infected) model is used to simulate potential disease transmission through the network between connected counties.
Monte Carlo simulations are then used to estimate county-level infection probabilities over a 5-step simulated timeline.

The project combines graph theory, probability modeling, computational simulation and geographic data visualization using Python.

## Stochastic Disease Spread Model

This model assumes that all infected neighbors are equally likely to spread infection to neighbors. This model is shown below:

$$P_{i}(t) = 1 - (1 - \beta)^{k_{i}(t)}$$

Where $P_{i}$ is the calculated probability of being infected, $\beta$ represents the ability of a disease to spread to neighboring counties, 
and $k_{i}(t)$ is the amount of neighboring counties infected.
This formula assumes that each infected neighboring county independenntly transmits infection to a susceptible county with probability $\beta$ during each simulated time step.

```python
#Create a function to be used stochastically to simulate disease spread through a network
def spread_simulation(G, initial_infected, spread_chance, timeframe, seed):
    '''Stochastic model to simulate disease spread through a network, based on a single spread chance parameter. Requires network, list of initially infected nodes,
    the chance of the disease spreading, the timeframe for the simulation to run on, and a seed to generate rng'''

    #Create rng set to be used
    rng = np.random.default_rng(seed=seed)

    #Use initial list of infected nodes to create a set
    infected = set(initial_infected)

    #Create a copy of initial infected to be stored in history
    timeline = [infected.copy()]

    #Iterate through desired time of simulations
    for step in range(timeframe):

        #Create a blank set to store newly infected nodes in
        new_infections = set()

        #Iterate through all nodes
        for county in G.nodes():

            #Skip county if it is already infected
            if county in infected:
                continue

            #Count how many connecting nodes are infected
            k= sum(neighbor in infected for neighbor in G.neighbors(county))

            #If no infected neighbors, assume no infection spread risk
            if k == 0:
                continue

            #Calculate infection chance using spread chance and number of neighbors
            infection_chance = 1 - (1 - spread_chance) ** k

            #Check against rng for stochastic simulation
            if rng.random() < infection_chance:

                #If rng check succeedes make the county infected
                new_infections.add(county)

        #Update infected list with new infections
        infected.update(new_infections)

        #Add infected list to timeline to track history
        timeline.append(infected.copy())

    return timeline
```

## Monte Carlo Simulation

### Initial Conditions
For the exploratory simulation, two counties are selected as initially infected counties to demonstrate how the stochastic simulation would
propagate through the network of counties. 

The current transmission parameter is set at $0.1$, which is a purely illustrative value.

These counties are hypothetical starting conditions and are not designated as historically accurate WNS locations. Future versions will
incorporate historical WNS surveillance records to establish initial infection locations, and create robust parameters to show likelihood of transmission between counties

```python
#Set parameters of 1000 simulations, 5 step timeframe, 2 arbitrary infected counties, and arbitrary spread chance
n_sims = 1000
timeframe = 5
initial_infected = [test_geoid, jeffco_geoid]
spread_chance = 0.1

#Create a list of all nodes in network
nodes = list(G.nodes)

#Create an array of zeros big enough to store all steps and all nodes
infection_counts = np.zeros((timeframe+1, len(nodes)))

#Create index to iterate through list with
node_index = {county: i for i, county in enumerate(nodes)}

#Create a blank array of zeros to store total infection numbers in for all steps
total_infected = np.zeros((n_sims, timeframe + 1))

#For every desired simulation use created function to model spread
for sim in range(n_sims):
    timeline = spread_simulation(G, initial_infected = initial_infected, spread_chance = spread_chance, timeframe = timeframe, seed = sim)

    for step, infected in enumerate(timeline):
        
        #Update total infected array with how many counties are infected in this simulation
        total_infected[sim, step] = len(infected)

        for county in infected:

            #If the county is in the infected list update the array at this county's postition to show in this simulation it was infected
            infection_counts[step, node_index[county]] += 1
```

```python
print(f'Number of simulations: {n_sims}')

print(f'Average infected counties by step: ' 
f'{total_infected.mean(axis = 0)}')
```

```python
#Generate Monte Carlo results using simulation to find counties frequently infected
infected_probability = infection_counts/n_sims

results = pd.DataFrame({
    'GEOID': nodes, 'Step 1': infected_probability[1], 'Step 2': infected_probability[2], 'Step 3': infected_probability[3],
    'Step 4': infected_probability[4], 'Step 5': infected_probability[5]
})
```

```python
#Identify counties with the highest simulated infection probability
results.sort_values(by = 'Step 5', ascending = False).head(25)
```

```python
#Calculate the mean of how many counties were infected for every sim in each step
mean_infected = total_infected.mean(axis=0)

print(mean_infected)

#Create central 90% simulation outcome interval
lower = np.percentile(total_infected, 5, axis=0)
upper = np.percentile(total_infected, 95, axis=0)

#Generate plot
plt.figure(figsize=(10, 6))

plt.plot(
    range(timeframe + 1),
    mean_infected,
    marker='.',
    label='Average infected counties'
)

#Fill between 90% simulation outcome interval
plt.fill_between(
    range(timeframe + 1),
    lower,
    upper,
    alpha=0.2,
    label='5th–95th percentile'
)

#Labels
plt.xlabel('Simulation Time Step')
plt.ylabel('Number of Infected Counties')
plt.title('WNS Stochastic Spread: 1,000 Simulations')
plt.legend()
plt.grid(alpha=0.3)
```

## Sensitivity Analysis

```python
#Create list of transmission variable values to test
sensitivities = [0.02, 0.05, 0.10, 0.20, 0.40]

#Set parameters of 1000 simulations, 5 step timeframe, 2 arbitrary infected counties
n_sims = 1000
timeframe = 5
initial_infected = [test_geoid, jeffco_geoid]
sensitivity_means = np.zeros((len(sensitivities), timeframe + 1))
sensitivity_lower = np.zeros((len(sensitivities), timeframe + 1))
sensitivity_upper = np.zeros((len(sensitivities), timeframe + 1))

sensitivity_record = []

for i, beta in enumerate(sensitivities):
    sensitivity_infected = np.zeros((n_sims, timeframe+1))
    for sim in range(n_sims):
        timeline = spread_simulation(G, initial_infected = initial_infected, spread_chance = beta, timeframe = timeframe, seed = sim) 
        for step, infected in enumerate(timeline):
            sensitivity_infected[sim][step] = len(infected)
    mean = np.mean(sensitivity_infected, axis = 0)
    sensitivity_means[i] = mean
    upper = np.percentile(sensitivity_infected, 95, axis=0)
    sensitivity_upper[i] = upper
    lower = np.percentile(sensitivity_infected, 5, axis=0)
    sensitivity_lower[i] = lower
    plt.plot(range(timeframe + 1), mean, '.-', label = f'Transmission Value = {beta}')
    plt.fill_between(range(timeframe + 1), lower, upper, alpha = 0.2)
    for step in range(timeframe+1):
        sensitivity_record.append({
            'Transmission Probability': beta,
            'Simulation Time Step': step,
            'Mean Infected Counties': sensitivity_means[i][step],
            '5th Percentile': sensitivity_lower[i, step],
            '95th Percentile': sensitivity_upper[i, step]
        })
plt.xlabel('Simulation Time Step')
plt.legend()
plt.ylabel('Mean Number of Infected Counties')
plt.title('Sensitivity Analysis of Stochastic WNS Spread')
plt.savefig('../outputs/figures/Sensitivity Analysis.png');
```

```python
sensitivity_results = pd.DataFrame(sensitivity_record)

display(sensitivity_results)
sensitivity_results.to_csv('../outputs/tables/WNS_Sensitivity_Analysis_Summary.csv', index = False)
```

```python
baseline = sensitivities.index(0.10)

print(f'Baseline mean: {total_infected.mean(axis = 0)}')
print(f'Sensitivity mean with beta = 0.1: {sensitivity_means[baseline]}')

assert np.allclose(
    total_infected.mean(axis=0),
    sensitivity_means[baseline]
)

print("Baseline and sensitivity results match.")
```

## Current Limitations

The current model exists as an exploratory stochastic network simulation with several limitations

1. Initial infection locations are hypothetical demonstration conditions rather than verified
    historical detections.
   
2. The transmission probability is assumed to be constant through all county connections.

3. Disease spread is limited to geographically adjacent counties.

4. Long-distance bat movement and other transmission pathways are not represented.

5. Environmental characteristics including cave density, bat habitat, winter climate, and others are not included.

6. The model does not distinguish between actual infection and recorded disease detection

7. Infected counties remain infected through the entire simulation

8. The transmission probability is an assumed parameter and has not been calibrated using historical WNS observations.

9. Simulation time steps do not currently correspond to a defined calendar duration.

10. County adjacency edge length is stored but not currently used for transmission calculations.

Future work will aim to incorporate historical WNS surveillance, ecological variables, and model calibration against historical observations.
